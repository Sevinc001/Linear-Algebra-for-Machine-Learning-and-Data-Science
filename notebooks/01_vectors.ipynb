# 01 — Vectors

This notebook builds vector operations from scratch (`src/vector_operations.py`)
and checks each one against NumPy's built-in equivalent.

Covered: vector addition, scalar multiplication, dot product, L1/L2 norms,
Euclidean distance, and cosine similarity.

```python
import sys
sys.path.insert(0, "../src")

import numpy as np
np.set_printoptions(precision=4, suppress=True)
```

## Define two example vectors

```python
u = np.array([1, 2, 3])
v = np.array([4, 5, 6])
u, v
```

## Vector addition and scalar multiplication

$$(\mathbf{u} + \mathbf{v})_i = u_i + v_i \qquad (c\mathbf{u})_i = c \cdot u_i$$

```python
from vector_operations import vector_add, vector_scale

print("u + v  (mine) :", vector_add(u, v))
print("u + v  (numpy):", u + v)
print()
print("2u     (mine) :", vector_scale(u, 2))
print("2u     (numpy):", 2 * u)
```

## Dot product

$$\mathbf{u} \cdot \mathbf{v} = \sum_i u_i v_i$$

The dot product measures how much two vectors point in the same direction,
and is the building block for almost everything else in this notebook —
norms, distances, projections, and (later) covariance.

```python
from vector_operations import dot_product

print("u . v (mine) :", dot_product(u, v))
print("u . v (numpy):", np.dot(u, v))
```

## Norms

The **L2 norm** (Euclidean length) and **L1 norm** (Manhattan length) of a
vector:

$$\|\mathbf{u}\|_2 = \sqrt{\sum_i u_i^2} \qquad \|\mathbf{u}\|_1 = \sum_i |u_i|$$

```python
from vector_operations import l2_norm, l1_norm

print("||u||_2 (mine) :", l2_norm(u))
print("||u||_2 (numpy):", np.linalg.norm(u))
print()
print("||u||_1 (mine) :", l1_norm(u))
print("||u||_1 (numpy):", np.linalg.norm(u, ord=1))
```

## Distance and similarity

**Euclidean distance** measures how far apart two vectors are in space.
**Cosine similarity** measures the angle between them, ignoring magnitude —
this is the metric behind most text/embedding similarity searches in ML.

```python
from vector_operations import euclidean_distance, cosine_similarity

print("distance(u, v)        :", euclidean_distance(u, v))
print("distance(u, v), numpy :", np.linalg.norm(u - v))
print()
print("cosine_similarity(u, v):", cosine_similarity(u, v))
```

## Why this matters for ML

- **Dot products** power similarity search, attention mechanisms, and linear layers.
- **Norms** are used in regularization (L1/L2 penalties) and gradient clipping.
- **Cosine similarity** is the standard metric for comparing embeddings
  (word vectors, sentence embeddings, recommendation systems).
