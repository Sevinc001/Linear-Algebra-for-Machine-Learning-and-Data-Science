## Define two example matrices

```python
A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])
A, B
```

## Matrix multiplication

$$(AB)_{ij} = \sum_k A_{ik} B_{kj}$$

Matrix multiplication is composition of linear transformations — this is
the single most-used operation in machine learning (every layer of a
neural network is, at its core, a matrix multiplication).

```python
from matrix_operations import matrix_multiply

print("A @ B (mine):\n", matrix_multiply(A, B))
print()
print("A @ B (numpy):\n", A @ B)
```

## Transpose

$$(A^T)_{ij} = A_{ji}$$

```python
from matrix_operations import transpose

print("A.T (mine):\n", transpose(A))
print()
print("A.T (numpy):\n", A.T)
```

## Identity matrix

$AI = A$ for any matrix $A$ — verified below.

```python
from matrix_operations import identity, matrix_multiply

I = identity(2)
print("I:\n", I)
print()
print("A @ I (should equal A):\n", matrix_multiply(A, I))
```

## Why this matters for ML

Matrix multiplication is how data flows through a neural network layer:
`output = activation(W @ x + b)`. Understanding it at the element level
demystifies backpropagation, which is fundamentally the chain rule applied
across a sequence of matrix multiplications.
